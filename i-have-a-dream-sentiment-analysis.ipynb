Here we imported the full text of Dr. Martin Luther King's I have a dream speech. To make it easy, I also imported the wordCount table of each word in the text.

Now we look at the sentiment polarity of the full text using TextBlob

```python
blob = TextBlob(text)
sentimentBlob = blob.sentiment
sentiment = [["Polarity",blob.sentiment.polarity],["Subjectivity",blob.sentiment.subjectivity]]
print(sentiment)

sentimentdf = pd.DataFrame(sentiment, columns=['Sentiment', 'Score'])
sentimentdf.plot(x="Sentiment", 
        y= "Score",
        kind="bar", 
        title = "Sentiment Score of 'I Have A Dream'",
        figsize=(10, 5))
```

Here we see that the speech's polarity is 0.14538889750976147

and its subjectivity is 0.46069990875137906

The polarity score is between -1 and 1 where -1 is the most negative and 1 is the most positive. 

The subjectivity score is between 0 and 1 where 0 is very objective and 1 is very subjective.

The speech's polarity is in the positive range scoring above 0.
